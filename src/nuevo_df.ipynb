```python
import pandas as pd
import os
```

```python
df_crimen = pd.read_csv('/workspaces/sp-ml-20-final-project-g4/data/processed/df_final (5).csv')
df_crimen
```

```python
df_madrid_p = pd.read_csv('../data/raw/Poblacion/poblacion-madrid.csv', sep=';', encoding='latin1')
df_barcelona_p = pd.read_csv('../data/raw/Poblacion/poblacion-Barcelona.csv', sep=';', encoding='latin1')
df_valencia_p = pd.read_csv('../data/raw/Poblacion/poblacion-valencia.csv', sep=';', encoding='latin1')
df_castellon_p = pd.read_csv('../data/raw/Poblacion/poblacion-castellon.csv', sep=';', encoding='latin1')
df_alicante_p = pd.read_csv('../data/raw/Poblacion/poblacion-alicante.csv', sep=';', encoding='latin1')


df_castellon_p = df_castellon_p.rename(columns={'Municipalities': 'Municipios', 'Sex': 'Sexo', 'Period': 'Periodo'})
df_alicante_p = df_alicante_p.rename(columns={'Municipalities': 'Municipios', 'Sex': 'Sexo', 'Period': 'Periodo'})
```

```python

df_alicante_p['Municipios'] = df_alicante_p['Municipios'].str.split('/').str[0].str.strip()
df_alicante_p['Municipios'] = df_alicante_p['Municipios'].str.replace(r'^\d+\s+', '', regex=True)
df_alicante_p['Municipios'] = df_alicante_p['Municipios'].replace('Alcoi', 'Alcoy')
df_alicante_p['Municipios'] = df_alicante_p['Municipios'].replace('Campello, el','Campello (el)' )
```

```python
# Definir los cambios: 'valor_viejo': 'valor_nuevo'
cambios_Alicane = {'Alcoi': 'Alcoy',
                   'Campello, el':'Campello (el)',
                   'Elx':'Elche', 
                   "Alfàs del Pi, l'":"L'Alfàs del Pi"}

df_alicante_p['Municipios'] = df_alicante_p['Municipios'].replace(cambios_Alicane)
```

```python
# Definir los cambios: 'valor_viejo': 'valor_nuevo'
cambios_Barcelona = {"08101 Hospitalet de Llobregat, L'":"Hospitalet de Llobregat (L')",
                     '08086 Franqueses del Vallès, Les':'Les Franqueses del Vallès',
                     '08118 Masnou, El':'Masnou (El)',
                     '08169 Prat de Llobregat, El':'Prat de Llobregat (El)'}

df_barcelona_p['Municipios'] = df_barcelona_p['Municipios'].replace(cambios_Barcelona)
```

```python
# Definir los cambios: 'valor_viejo': 'valor_nuevo'
cambios_Madrid = {'28127 Rozas de Madrid, Las':'Las Rozas de Madrid'}

df_madrid_p['Municipios'] = df_madrid_p['Municipios'].replace(cambios_Madrid)
```

```python
# Definir los cambios: 'valor_viejo': 'valor_nuevo'
cambios_Valencia = {'46171 Montcada/Moncada':'Moncada',
                    '46202 Pobla de Vallbona, la':'Pobla de Vallbona (la)',
                    '46220 Sagunt/Sagunto':'Sagunto'}

df_valencia_p['Municipios'] = df_valencia_p['Municipios'].replace(cambios_Valencia)
```

```python
# Definir los cambios: 'valor_viejo': 'valor_nuevo'
cambios_Castellon = {"12126 Vall d'Uixó, la":"Vall d'Uixó (la)",
                     '12135 Vila-real':'Vila-Real'}

df_castellon_p['Municipios'] = df_castellon_p['Municipios'].replace(cambios_Castellon)
```

```python

df_pob = pd.concat([df_madrid_p, df_barcelona_p, df_valencia_p, df_castellon_p,df_alicante_p])
df_pob
```

```python
df_crimen = df_crimen.rename(columns={'Periodo': 'Periodo Trimestral'})
df_pob = df_pob.rename(columns={'Total': 'Poblacion'})
```

```python
df_pob['Poblacion'] = df_pob['Poblacion'].astype(str).str.replace('.', '', regex=False)
```

```python

df_pob['Municipio_Clean'] = df_pob['Municipios'].str.replace(r'^\d+\s+', '', regex=True)
```

```python
df_pob['Municipio_Clean'] = df_pob['Municipio_Clean'].str.split('/').str[0].str.strip()
df_crimen['Municipio'] = df_crimen['Municipio'].str.split('/').str[0].str.strip()
df_crimen['Municipio'] = df_crimen['Municipio'].str.strip()
```

```python
df_pob
```

```python
df_pob = df_pob.drop(columns=['Municipios'])
df_pob
```

```python
df_crimen
```

```python
# Unimos por Municipio_Clean y por el Año (que en df_pob se llama 'Periodo')
df_merged = df_crimen.merge(df_pob[['Municipio_Clean', 'Periodo', 'Poblacion']], 
                            left_on=['Municipio', 'Año'], 
                            right_on=['Municipio_Clean', 'Periodo'], 
                            how='left')
df_merged
```

```python
df_merged = df_merged.drop(columns=['Municipio_Clean', 'Periodo'])
df_merged
```
