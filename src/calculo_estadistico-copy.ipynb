```python
def calculo_estadistico_delitos(df):
    df = df.copy()

    lista_delitos = ["Delitos Sexuales",
                     "Homicidios",
                     "Hurtos",
                     "Lesiones",
                     "Resto De Criminalidad Convencional",
                     "Robos Con Fuerza",
                     "Robos Con Violencia",
                     "Secuestro",
                     "Sustraccion De Vehiculos",
                     "Trafico De Drogas",
                     "Total Delitos"]
   
    df = df.groupby("Geografía")[lista_delitos].mean()

    return df.reset_index()

media_delitos = calculo_estadistico_delitos(df)
media_delitos
```
